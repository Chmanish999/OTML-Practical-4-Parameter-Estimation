## What are Parameters?

Parameters are internal values or rules learned by a machine learning model from data. These values help the model make predictions.

Parameter estimation means finding the best parameter values or decision rules using training data. The goal is to achieve better prediction with minimum error.

| Model | Parameters |
|---|---|
| Linear Regression | Weights and bias |
| Decision Tree | Split rules |
| Neural Network | Weights and biases |
| SVM | Hyperplane coefficients |

````python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
````

````python
df = pd.read_csv("music1.csv")
df.head()
````

````python
X = df[["age", "gender"]]
y = df["gener"]

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)
model
````

````python
plt.figure(figsize=(12, 8))
plot_tree(
    model,
    feature_names=["age", "gender"],
    class_names=model.classes_,
    filled=True
)
plt.show()
````

````python
sample = pd.DataFrame([[22, 1]], columns=["age", "gender"])
prediction = model.predict(sample)
print("Predicted Genre:", prediction[0])
````

## Observation

During training, the machine learning model estimates internal decision rules from the data. These estimated rules act as parameters and help the model make predictions on new inputs.

The workflow is:

```text
Data → Learning → Parameter Estimation → Prediction
```
